```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Fungsi untuk mencetak isi list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # 1. Penyisipan di Awal (image_349855.png)
    def AddBegining(self, newdata):
        NewNode = Node(newdata)
        NewNode.nextval = self.headval
        self.headval = NewNode

    # 2. Penyisipan di Tengah/Setelah Node Tertentu (image_349f3d.png)
    def AddInbetween(self, mid_node, newdata):
        if mid_node is None:
            print("The mentioned node is absent")
            return

        NewNode = Node(newdata)
        NewNode.nextval = mid_node.nextval
        mid_node.nextval = NewNode

# Inisialisasi awal
Li = LinkedList()
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

Li.headval = n1
n1.nextval = n2
n2.nextval = n3

# Eksekusi fungsi
Li.AddBegining("Start")            # Menambah di awal list
Li.AddInbetween(n1, "Middle")      # Menambah setelah n1 ("Januari")
Li.listprint()
```

MENGHAPUS NODE PADA LINKED LIST


```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    def AddBegining(self, newdata):
        NewNode = Node(newdata)
        NewNode.nextval = self.headval
        self.headval = NewNode

    # Fungsi untuk menghapus node berdasarkan kunci (image_34aaa1.png)
    def RemoveNode(self, Removekey):
        HeadVal = self.headval

        if (HeadVal is not None):
            if (HeadVal.dataval == Removekey):
                self.headval = HeadVal.nextval
                HeadVal = None
                return

        while (HeadVal is not None):
            if HeadVal.dataval == Removekey:
                break
            prev = HeadVal
            HeadVal = HeadVal.nextval

        if (HeadVal == None):
            return

        prev.nextval = HeadVal.nextval
        HeadVal = None

# Eksekusi Kode
Li = LinkedList()
Li.AddBegining("April")
Li.AddBegining("Maret")
Li.AddBegining("Febuari")

print("Sebelum dilakukan remove")
Li.listprint()
print("---------------------------------------")

# Menghapus node dengan kunci "Maret"
Li.RemoveNode("Maret")
print("Setelah dilakukan remove dengan kunci 'Maret'")
Li.listprint()
```

```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Print the linked list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # Function to add node in the beginning
    def AddBegining(self, newdata):
        NewNode = Node(newdata)
        # Update the new nodes next val to existing node
        NewNode.nextval = self.headval
        self.headval = NewNode

    # Function to remove first node
    def RemoveFirst(self):
        afterhead = self.headval
        if afterhead is not None:
            self.headval = afterhead.nextval

# --- Eksekusi Program ---

# create linked list named Li
Li = LinkedList()

# add node
Li.AddBegining("April")
Li.AddBegining("Maret")
Li.AddBegining("Febuari")

print("Sebelum dilakukan remove")
Li.listprint()
print("---------------------------------------------------------")

# remove first node
Li.RemoveFirst()

print("Setelah dilakukan remove node pertama")
Li.listprint()
```

Hapus node di akhir

```python
class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Print the linked list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # Function to add newnode in the end
    def AddInEnd(self, newdata):
        NewNode = Node(newdata)
        if self.headval is None:
            self.headval = NewNode
            return
        last = self.headval
        while(last.nextval):
            last = last.nextval
        last.nextval = NewNode

    # Function to remove last node
    def RemoveEnd(self):
        last = self.headval
        while (last is not None):
            if last.nextval == None:
                break
            prev = last
            last = last.nextval

        if (last == None):
            return

        prev.nextval = last.nextval
        last = None

# --- Eksekusi Program ---

# create linked list named Li
Li = LinkedList()

# add node
Li.AddInEnd("Januari")
Li.AddInEnd("Februari")
Li.AddInEnd("Maret")

print("Sebelum dilakukan remove")
Li.listprint()
print("=========================================================")

# remove last node
Li.RemoveEnd()

print("Setelah dilakukan remove node terakhir")
Li.listprint()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoubleLinkedList:
    def __init__(self):
        self.head = None

    def display(self):
        curr = self.head
        elements = []
        while curr:
            elements.append(str(curr.data))
            curr = curr.next
        print(" <-> ".join(elements))

    # memasukkan data setelah node tertentu
    def insert_after(self, target_node_data, new_data):
        curr = self.head
        while curr:
            if curr.data == target_node_data:
                new_node = Node(new_data)
                new_node.next = curr.next
                new_node.prev = curr
                if curr.next:
                    curr.next.prev = new_node
                curr.next = new_node
                return
            curr = curr.next
        print(f"Node {target_node_data} tidak ditemukan.")

    # Hapus dengan kata kunci
    def delete_by_key(self, key):
        curr = self.head
        while curr:
            if curr.data == key:
                if curr.prev:
                    curr.prev.next = curr.next
                else:
                    self.head = curr.next

                if curr.next:
                    curr.next.prev = curr.prev
                curr = None
                return
            curr = curr.next
        print(f"Kata kunci '{key}' tidak ditemukan.")

# Eksekusi
dll = DoubleLinkedList()
# Membuat 3 node awal
n1 = Node("Januari")
n2 = Node("Maret")
n3 = Node("April")
dll.head = n1
n1.next = n2; n2.prev = n1
n2.next = n3; n3.prev = n2

print("List Awal (3 Node):")
dll.display()

print("\nPenyisipan 'Februari' setelah 'Januari':")
dll.insert_after("Januari", "Februari")
dll.display()

print("\nPenghapusan dengan kata kunci 'Maret':")
dll.delete_by_key("Maret")
dll.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def display(self):
        if not self.head: return
        curr = self.head
        while True:
            print(curr.data, end=" -> ")
            curr = curr.next
            if curr == self.head: break
        print("(kembali ke awal)")

    # Sisip Akhir
    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
            return
        curr = self.head
        while curr.next != self.head:
            curr = curr.next
        curr.next = new_node
        new_node.next = self.head

    # Hapus Akhir
    def remove_end(self):
        if not self.head: return
        if self.head.next == self.head:
            self.head = None
            return
        curr = self.head
        prev = None
        while curr.next != self.head:
            prev = curr
            curr = curr.next
        prev.next = self.head

# Eksekusi
cll = CircularLinkedList()
# Membuat 3 node awal
cll.insert_end("Node1")
cll.insert_end("Node2")
cll.insert_end("Node3")

print("Circular List Awal:")
cll.display()

print("\nPenyisipan Akhir 'Node4':")
cll.insert_end("Node4")
cll.display()

print("\nPenghapusan Akhir:")
cll.remove_end()
cll.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoubleLinkedList:
    def __init__(self):
        self.head = None

    # Tampilkan List
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" <-> ")
            curr = curr.next
        print("Kembali ke awal")

    # Sisip Tengah (Setelah node tertentu)
    def insert_after(self, target_data, new_data):
        curr = self.head
        while curr:
            if curr.data == target_data:
                new_node = Node(new_data)
                new_node.next = curr.next
                new_node.prev = curr
                if curr.next:
                    curr.next.prev = new_node
                curr.next = new_node
                return
            curr = curr.next

    # Hapus dengan Kata Kunci
    def delete_by_key(self, key):
        curr = self.head
        while curr:
            if curr.data == key:
                if curr.prev: curr.prev.next = curr.next
                else: self.head = curr.next
                if curr.next: curr.next.prev = curr.prev
                return
            curr = curr.next

# Eksekusi
dll = DoubleLinkedList()
n1, n2, n3 = Node("A"), Node("B"), Node("C")
dll.head = n1
n1.next = n2; n2.prev = n1
n2.next = n3; n3.prev = n2

dll.insert_after("B", "X") # Sisip X setelah B
dll.delete_by_key("C")      # Hapus C
dll.display() # Hasil: A <-> B <-> X <->
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def display(self):
        if not self.head: return
        curr = self.head
        while True:
            print(curr.data, end=" -> ")
            curr = curr.next
            if curr == self.head: break
        print("(Back to Head)")

    # Sisip Akhir
    def add_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
            return
        curr = self.head
        while curr.next != self.head:
            curr = curr.next
        curr.next = new_node
        new_node.next = self.head

    # Hapus Akhir
    def remove_end(self):
        if not self.head or self.head.next == self.head:
            self.head = None
            return
        curr = self.head
        while curr.next.next != self.head:
            curr = curr.next
        curr.next = self.head

# Eksekusi
cll = CircularLinkedList()
cll.add_end("1")
cll.add_end("2")
cll.add_end("3") # 3 Node awal

cll.add_end("4") # Sisip akhir
cll.remove_end() # Hapus akhir
cll.display()    # Hasil: 1 -> 2 -> 3 -> (Back to Head)
```
